```python
from google.colab import output
from IPython.display import display, Javascript
import cv2
import numpy as np
from PIL import Image
import io
import base64
from fer import FER

detector = FER(mtcnn=True)

def take_photo(filename='photo.jpg', quality=0.8):
    js = Javascript('''
    async function takePhoto(quality) {
      const div = document.createElement('div');
      const video = document.createElement('video');
      const button = document.createElement('button');
      button.textContent = 'Capture';
      div.appendChild(video);
      div.appendChild(button);
      document.body.appendChild(div);

      const stream = await navigator.mediaDevices.getUserMedia({video: true});
      video.srcObject = stream;
      await video.play();

      // Wait for user to click
      await new Promise((resolve) => button.onclick = resolve);

      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;
      canvas.getContext('2d').drawImage(video, 0, 0);
      stream.getTracks().forEach(track => track.stop());
      div.remove();
      return canvas.toDataURL('image/jpeg', quality);
    }
    ''')
    display(js)
    data = output.eval_js('takePhoto(%f)' % quality)

    # Convert base64 to OpenCV image
    content = data.split(',')[1]
    decoded = base64.b64decode(content)
    img = Image.open(io.BytesIO(decoded))
    img = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)
    cv2.imwrite(filename, img)
    return img

# use
img = take_photo()
emotion, score = detector.top_emotion(img)
print(f"Emotion: {emotion}, Score: {score:.2f}")
```

```python
from google.colab import output
from IPython.display import display, Javascript
import cv2
import numpy as np
from PIL import Image
import io
import base64
from fer import FER
import time
from google.colab.patches import cv2_imshow  # Colab-friendly imshow

# Initialize FER emotion detector with MTCNN
detector = FER(mtcnn=True)

# ==========================
# Function: Capture Photo
# ==========================
def take_photo(quality=0.8):
    """
    Opens webcam in Colab and captures a single snapshot.
    Returns: BGR image (numpy array)
    """
    js = Javascript('''
    async function takePhoto(quality) {
      const div = document.createElement('div');
      const video = document.createElement('video');
      const button = document.createElement('button');
      button.textContent = 'Capture';
      div.appendChild(video);
      div.appendChild(button);
      document.body.appendChild(div);

      const stream = await navigator.mediaDevices.getUserMedia({video: true});
      video.srcObject = stream;
      await video.play();

      await new Promise((resolve) => button.onclick = resolve);

      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;
      canvas.getContext('2d').drawImage(video, 0, 0);
      stream.getTracks().forEach(track => track.stop());
      div.remove();
      return canvas.toDataURL('image/jpeg', quality);
    }
    ''')
    display(js)
    data = output.eval_js('takePhoto(%f)' % quality)
    content = data.split(',')[1]
    decoded = base64.b64decode(content)
    img = Image.open(io.BytesIO(decoded))
    img = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)
    return img

# ==========================
# Capture Multiple Snapshots & Detect Emotions
# ==========================
frames = []
num_shots = 5

for i in range(num_shots):
    print(f"Take snapshot {i+1}/{num_shots}")
    img = take_photo()

    # Detect dominant emotion
    emotion, score = detector.top_emotion(img)
    if emotion is None or score is None:
        print("No face detected")
        emotion = "Unknown"
        score = 0.0
    else:
        print(f"Detected: {emotion} ({score:.2f})")

    frames.append((img, emotion))
    time.sleep(1)  # Wait 1 second between snapshots

# ==========================
# Display Snapshots & Emotions
# ==========================
for idx, (img, emo) in enumerate(frames):
    cv2_imshow(img)  # Show image in Colab
    print(f"Snapshot {idx+1}: Emotion = {emo}")

cv2.destroyAllWindows()
```

```python
from google.colab import output
from IPython.display import display, Javascript
import cv2
import numpy as np
from PIL import Image
import io
import base64
from fer import FER
import time
from google.colab.patches import cv2_imshow  # Colab-compatible imshow

# Initialize FER detector with MTCNN
detector = FER(mtcnn=True)

# ==========================
# Function: Capture Photo from Webcam
# ==========================
def take_photo(quality=0.8):
    """
    Opens webcam in Colab and captures a single snapshot.
    Returns: BGR image (numpy array)
    """
    js = Javascript('''
    async function takePhoto(quality) {
      const div = document.createElement('div');
      const video = document.createElement('video');
      const button = document.createElement('button');
      button.textContent = 'Capture';
      div.appendChild(video);
      div.appendChild(button);
      document.body.appendChild(div);

      const stream = await navigator.mediaDevices.getUserMedia({video: true});
      video.srcObject = stream;
      await video.play();

      await new Promise((resolve) => button.onclick = resolve);

      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;
      canvas.getContext('2d').drawImage(video, 0, 0);
      stream.getTracks().forEach(track => track.stop());
      div.remove();
      return canvas.toDataURL('image/jpeg', quality);
    }
    ''')
    display(js)
    data = output.eval_js('takePhoto(%f)' % quality)
    content = data.split(',')[1]
    decoded = base64.b64decode(content)
    img = Image.open(io.BytesIO(decoded))
    img = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)
    return img

# ==========================
# Capture Multiple Snapshots and Detect Emotions
# ==========================
frames = []
num_shots = 5

for i in range(num_shots):
    print(f"Taking snapshot {i+1}/{num_shots}")
    img = take_photo()

    # Detect dominant emotion with FER
    emotion, score = detector.top_emotion(img)
    if (emotion is None or score is None) or (score < 0.5):
        print("No face detected")
        emotion = "Unknown"
        score = 0.0
    else:
        print(f"Detected: {emotion} ({score:.2f})")

    frames.append((img, emotion))
    time.sleep(1)  # Wait 1 second between snapshots

# ==========================
# Display Snapshots and Emotions
# ==========================
for idx, (img, emo) in enumerate(frames):
    cv2_imshow(img)  # Display image in Colab
    print(f"Snapshot {idx+1}: Emotion = {emo}")

cv2.destroyAllWindows()
```
