```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import Adam

# keras-rl2
from rl.agents.dqn import DQNAgent
from rl.policy import EpsGreedyQPolicy
from rl.memory import SequentialMemory

# Custom NAO 6-emotion environment (must implement Gym interface)
from nao_env_6emo import Nao6EmotionEnv

# ==============================
# 1. Create Environment
# ==============================
env = Nao6EmotionEnv()
np.random.seed(42)
tf.random.set_seed(42)
env.seed(42)

nb_actions = env.action_space.n  # จำนวน action ของ environment

# ==============================
# 2. Build Neural Network Model
# ==============================
model = Sequential()
model.add(Flatten(input_shape=(1,) + env.observation_space.shape))
model.add(Dense(64, activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(nb_actions, activation='linear'))

print(model.summary())

# ==============================
# 3. Configure DQN Agent
# ==============================
memory = SequentialMemory(limit=50000, window_length=1)
policy = EpsGreedyQPolicy(eps=0.1)  # exploration

dqn = DQNAgent(model=model, nb_actions=nb_actions, memory=memory, nb_steps_warmup=1000,
               target_model_update=0.01, policy=policy, gamma=0.99, batch_size=32)
dqn.compile(Adam(learning_rate=1e-3), metrics=['mae'])

# ==============================
# 4. Train Agent
# ==============================
print("Start training DQN agent...")
dqn.fit(env, nb_steps=20000, visualize=False, verbose=2)

# ==============================
# 5. Save Model
# ==============================
model_name = "nao_6emo_dqn"
dqn.save_weights(model_name + ".h5f", overwrite=True)
print(f"Training finished. Model saved as {model_name}.h5f")

# ==============================
# 6. Test / Evaluate Agent
# ==============================
print("Testing agent for 5 episodes...")
dqn.test(env, nb_episodes=5, visualize=True)
```

```python
import gymnasium as gym
from gymnasium import spaces
import numpy as np
import random

# กำหนดชื่ออารมณ์และแอ็กชันเพื่อให้โค้ดอ่านง่าย (สำหรับรายงาน)
# 0:Neutral/Patrol, 1:Happy/Wave, 2:Angry/Stomp, 3:Sad/Slouch, 4:Surprise/TaiChi
EMOTION_LABELS = {
    0: "Neutral",
    1: "Happy",
    2: "Angry",
    3: "Sad",
    4: "Surprise"
}

class NaoDecisionEnv(gym.Env):
    # กำหนด metadata ตามหลักปฏิบัติของ Gymnasium
    metadata = {"render_modes": ["human", "rgb_array"], "render_fps": 30}

    def __init__(self, render_mode=None):
        super(NaoDecisionEnv, self).__init__()

        # --- Space Definition ---
        # Action space: AI choose from 5 actions (0 to 4)
        self.action_space = spaces.Discrete(len(EMOTION_LABELS)) # ขนาด = 5

        # Observation space: AI sees 5 emotions (0 to 4)
        # Use spaces.Discrete to define the number of states.
        self.observation_space = spaces.Discrete(len(EMOTION_LABELS)) # ขนาด = 5

        # --- Environment State ---
        # current_obs จะเก็บค่า int (0-4) ที่เป็นตัวแทนของอารมณ์ที่สังเกตได้
        self.current_obs = None
        self.episode_length = 50
        self.steps_left = self.episode_length
        self.render_mode = render_mode
        self.action_labels = EMOTION_LABELS # อ้างอิงเพื่อให้เรียกใช้ได้ง่าย

    def reset(self, seed=None, options=None):
        """รีเซ็ตสภาพแวดล้อมเพื่อเริ่มต้น Episode ใหม่"""
        super().reset(seed=seed)

        # สุ่มอารมณ์เริ่มต้น (0-4)
        # ใช้ np.int64 เพื่อให้เข้ากันได้กับ RL libraries เช่น Stable-Baselines3
        self.current_obs = np.int64(self.np_random.randint(0, len(EMOTION_LABELS)))
        self.steps_left = self.episode_length

        # info dict สามารถใช้ส่งข้อมูลเพิ่มเติมที่ไม่ใช่ state (เช่น ชื่ออารมณ์)
        info = {"emotion_name": EMOTION_LABELS[self.current_obs]}

        return self.current_obs, info

    def step(self, action):
        """ดำเนินการ 1 Step ในสภาพแวดล้อม"""
        reward = 0.0

        # --- Reward Function (Key Point) ---
        # Note: การออกแบบ Reward function เป็นหัวใจสำคัญของรายงาน

        # 1. Successful Match (High Reward)
        if action == self.current_obs:
            reward = 10.0
            # เพิ่มความชัดเจนของชื่ออารมณ์/แอ็กชันในรายงาน
            # print(f"✅ Correct! Obs: {EMOTION_LABELS[self.current_obs]}, Action: {EMOTION_LABELS[action]}")

        # 2. Mild Mismatch (Small Penalty) - เมื่อแอ็กชันใกล้เคียงอารมณ์จริง
        # ตัวอย่าง: โมเดลเลือก Happy (1) แต่คนทำ Surprise (4)
        elif abs(action - self.current_obs) == 1:
             reward = -1.0 # ลงโทษเล็กน้อย

        # 3. Complete Mismatch (High Penalty) - เมื่อแอ็กชันผิดพลาดอย่างมาก
        else:
            reward = -5.0 # ลงโทษปานกลางถึงสูง
            # print(f"❌ Mismatch. Obs: {EMOTION_LABELS[self.current_obs]}, Action: {EMOTION_LABELS[action]}")

        # --- Transition ---
        # สุ่มเปลี่ยนไปอารมณ์ถัดไป (สำหรับกรณีที่โมเดลต้องเรียนรู้การตอบสนองต่ออารมณ์ที่เปลี่ยนไปอย่างรวดเร็ว)
        self.current_obs = np.int64(self.np_random.randint(0, len(EMOTION_LABELS)))

        # --- Termination Check ---
        self.steps_left -= 1
        terminated = self.steps_left <= 0
        truncated = False # ในสภาพแวดล้อมนี้ไม่มีการ Truncation

        # info dict
        info = {"emotion_name": EMOTION_LABELS[self.current_obs]}

        # Gymnasium/RL requires 5 return values
        return self.current_obs, reward, terminated, truncated, info

    def render(self):
        # ฟังก์ชัน render จะเว้นว่างไว้สำหรับหุ่นยนต์ NAO (ใน Webots)
        pass

    def close(self):
        # ฟังก์ชัน close จะเว้นว่างไว้สำหรับการรันในคอนโซล
        pass
```

```python
"""

Note: Simple Q-Learning implementation for NAO robot's emotion response.
This script uses a classic, table-based RL approach without the Stable-Baselines3 library.
"""

import numpy as np
import random
import os
import time

# --- 1. Environment and Constants Setup ---

# Define the number of states (emotions) and actions (robot responses)
# State 0:Neutral, 1:Happy, 2:Angry, 3:Sad, 4:Surprise
NUM_STATES = 5
# Action 0:Patrol, 1:Wave, 2:Stomp, 3:Slouch, 4:TaiChi
NUM_ACTIONS = 5

# Learning Parameters for Q-Learning
LEARNING_RATE = 0.1   # Alpha: How quickly the agent accepts new information (e.g., 0.1 means 10% new info)
DISCOUNT_FACTOR = 0.9 # Gamma: Importance of future rewards (closer to 1.0 means more far-sighted)
NUM_EPISODES = 5000   # Number of training rounds (episodes)

# Epsilon-Greedy Strategy Parameters (Controls Exploration vs. Exploitation)
EPSILON_START = 1.0   # Start with 100% exploration (random actions)
EPSILON_END = 0.01    # End with 1% exploration (mostly using learned knowledge)
EPSILON_DECAY_RATE = 0.0005 # Rate at which exploration decreases over episodes

# Reward Definition (Based on the environment logic: matching emotion = high reward)
REWARD_MATCH = 10.0
REWARD_MISMATCH_MINOR = -1.0 # Small penalty for close-but-wrong action
REWARD_MISMATCH_MAJOR = -5.0 # High penalty for completely wrong action

# File name for saving the final learned Q-Table
MODEL_FILENAME = "nao_emotion_qtable.npy"

# --- 2. Q-Learning Core Functions (No Class Structure) ---

def get_reward(state, action):
    """Calculates the reward based on the observed emotion (state) and the robot's action."""
    if action == state:
        return REWARD_MATCH  # High reward for correct response
    elif abs(action - state) == 1:
        return REWARD_MISMATCH_MINOR # Small penalty for minor error
    else:
        return REWARD_MISMATCH_MAJOR # High penalty for major error

def choose_action(state, q_table, epsilon):
    """
    Chooses the next action using the Epsilon-Greedy Strategy.
    The agent flips a coin: choose randomly (explore) or choose the best Q-value (exploit).
    """
    if random.random() < epsilon:
        # Explore: Choose a random action
        return random.randint(0, NUM_ACTIONS - 1)
    else:
        # Exploit: Choose the action with the highest predicted reward from the Q-Table
        # np.argmax returns the index (action) of the maximum Q-value for the current state
        return np.argmax(q_table[state, :])

def run_episode(q_table, epsilon):
    """
    Runs a single training step (or episode for this simple problem) and updates the Q-Table.
    This function implements the core Q-Learning update rule.
    """
    # 1. Initialize State: Start a new state (emotion) randomly
    current_state = random.randint(0, NUM_STATES - 1)

    # 2. Agent chooses an action
    action = choose_action(current_state, q_table, epsilon)

    # 3. Environment interaction: Get reward and next state
    reward = get_reward(current_state, action)
    # In this simple model, the "next state" is just a new random emotion for the next decision
    next_state = random.randint(0, NUM_STATES - 1)

    # 4. Q-Table Update using the Bellman Equation
    # New Q(s,a) = Old Q(s,a) + LR * (Reward + Gamma * Max(Q(s',a')) - Old Q(s,a))

    # Max(Q(s',a')) is the maximum possible future reward from the next state (s')
    max_future_q = np.max(q_table[next_state, :])

    # The actual Q-Learning update formula
    old_q_value = q_table[current_state, action]
    new_q_value = old_q_value + LEARNING_RATE * (reward + DISCOUNT_FACTOR * max_future_q - old_q_value)

    # Update the table
    q_table[current_state, action] = new_q_value

    return q_table, reward


# --- 3. Main Training Script ---

def train_q_learning_agent():
    """Initializes and trains the Q-Learning agent."""

    print("Initializing Q-Table...")
    # Initialize the Q-Table (State x Action matrix) with zeros
    q_table = np.zeros((NUM_STATES, NUM_ACTIONS))

    # Calculate how much epsilon (exploration) should decrease in each episode
    epsilon_decay_values = np.linspace(EPSILON_START, EPSILON_END, NUM_EPISODES)

    print(f"Starting Q-Learning training for {NUM_EPISODES} episodes...")

    start_time = time.time()

    # Main loop that iterates through all training episodes
    for episode in range(NUM_EPISODES):
        # Get the current epsilon value for the Epsilon-Greedy strategy
        epsilon = epsilon_decay_values[episode]

        # Run the training step
        q_table, reward = run_episode(q_table, epsilon)

        # Print a progress update every 500 episodes
        if (episode + 1) % 500 == 0:
            # Calculate average value of the best action for all states
            avg_max_q = np.mean(np.max(q_table, axis=1))
            print(f"Episode: {episode + 1}/{NUM_EPISODES} | Epsilon: {epsilon:.4f} | Avg Max Q: {avg_max_q:.2f}")

    end_time = time.time()

    print("\n--- Training Complete ---")
    print(f"Total time taken: {end_time - start_time:.2f} seconds")
    print("Final Learned Q-Table Analysis:")

    # --- Analysis of the Final Q-Table ---
    emotion_labels = ["Neutral", "Happy", "Angry", "Sad", "Surprise"]
    action_labels = ["Patrol", "Wave", "Stomp", "Slouch", "TaiChi"]

    print("------------------------------------------------------------------")
    print(f"{'State (Emotion)':<20} | {'Optimal Action':<15} | {'Q-Values (Action 0 to 4)':<}")
    print("------------------------------------------------------------------")

    for state in range(NUM_STATES):
        # Find the best action for the current state (the highest Q-value)
        best_action_index = np.argmax(q_table[state, :])

        # Format Q-values for clear display
        q_values_formatted = [f"{q:.2f}" for q in q_table[state, :]]
        q_values_str = " | ".join(q_values_formatted)

        print(f"{emotion_labels[state]:<20} | {action_labels[best_action_index]} ({best_action_index}) | {q_values_str}")

    print("------------------------------------------------------------------")

    # Save the Q-Table to a file for the Webots controller to load
    np.save(MODEL_FILENAME, q_table)
    print(f"Q-Table saved successfully as: {MODEL_FILENAME}")
    print("NEXT STEP: You need to copy this saved NPY file into your Webots controller folder!")


if __name__ == "__main__":
    train_q_learning_agent()
```

```python
"""

Note: Script for Hyperparameter Tuning (Grid Search) of the Q-Learning agent.
This finds the best combination of Learning Rate (Alpha) and Discount Factor (Gamma).
"""

import numpy as np
import random
import time
import os
import itertools # Used to generate all combinations of parameters for Grid Search

# --- 1. Environment and Constants Setup ---

# Define the number of states (emotions) and actions (robot responses)
NUM_STATES = 5
NUM_ACTIONS = 5

# Training Constants (These will remain fixed during tuning)
NUM_EPISODES = 2000   # Reduced episodes for faster tuning, but can be increased later
MODEL_FILENAME = "nao_emotion_qtable_tuned.npy"

# Reward Definition (Based on the environment logic)
REWARD_MATCH = 10.0
REWARD_MISMATCH_MINOR = -1.0
REWARD_MISMATCH_MAJOR = -5.0

# Hyperparameters to test (Grid Search space)
# We will test all combinations of these values
ALPHA_VALUES = [0.05, 0.1, 0.2] # Learning Rate (Alpha)
GAMMA_VALUES = [0.8, 0.9, 0.99] # Discount Factor (Gamma)

# Epsilon-Greedy Strategy Parameters
EPSILON_START = 1.0
EPSILON_END = 0.01
# The decay rate will be calculated inside the training function based on NUM_EPISODES

# --- 2. Q-Learning Core Functions (Helper Functions) ---

def get_reward(state, action):
    """Calculates the reward based on the observed emotion (state) and the robot's action."""
    if action == state:
        return REWARD_MATCH
    elif abs(action - state) == 1:
        return REWARD_MISMATCH_MINOR
    else:
        return REWARD_MISMATCH_MAJOR

def choose_action(state, q_table, epsilon):
    """
    Chooses the next action using the Epsilon-Greedy Strategy.
    """
    if random.random() < epsilon:
        # Explore: Choose a random action
        return random.randint(0, NUM_ACTIONS - 1)
    else:
        # Exploit: Choose the action with the highest predicted reward
        return np.argmax(q_table[state, :])

def train_agent(learning_rate, discount_factor, num_episodes):
    """
    Trains the Q-Learning agent with a specific set of hyperparameters.
    Returns the final Q-Table and the total cumulative reward for evaluation.
    """
    q_table = np.zeros((NUM_STATES, NUM_ACTIONS))
    total_reward = 0

    # Calculate epsilon decay values based on the specific number of episodes
    epsilon_decay_values = np.linspace(EPSILON_START, EPSILON_END, num_episodes)

    for episode in range(num_episodes):
        current_state = random.randint(0, NUM_STATES - 1)
        epsilon = epsilon_decay_values[episode]

        action = choose_action(current_state, q_table, epsilon)
        reward = get_reward(current_state, action)
        next_state = random.randint(0, NUM_STATES - 1)

        # Q-Table Update using the Bellman Equation
        max_future_q = np.max(q_table[next_state, :])
        old_q_value = q_table[current_state, action]

        new_q_value = old_q_value + learning_rate * (reward + discount_factor * max_future_q - old_q_value)
        q_table[current_state, action] = new_q_value

        total_reward += reward

    # The evaluation metric for tuning is the final average Max Q-Value,
    # which reflects the maximum potential reward the agent has learned.
    final_metric = np.mean(np.max(q_table, axis=1))

    return q_table, final_metric, total_reward

# --- 3. Hyperparameter Tuning Function (Grid Search) ---

def grid_search_tuning():
    """
    Performs Grid Search over predefined Alpha and Gamma values
    to find the optimal combination.
    """
    best_score = -np.inf
    best_params = {}
    best_q_table = None

    # itertools.product creates all combinations: (0.05, 0.8), (0.05, 0.9), ..., (0.2, 0.99)
    param_combinations = list(itertools.product(ALPHA_VALUES, GAMMA_VALUES))
    total_runs = len(param_combinations)

    print(f"--- Starting Grid Search Tuning ({total_runs} total runs) ---")

    for i, (alpha, gamma) in enumerate(param_combinations):
        print(f"\n[RUN {i+1}/{total_runs}] Training with: Alpha={alpha}, Gamma={gamma}")

        # Train the agent with the current set of parameters
        q_table, final_metric, total_reward = train_agent(
            learning_rate=alpha,
            discount_factor=gamma,
            num_episodes=NUM_EPISODES
        )

        print(f"  -> FINAL METRIC (Avg Max Q-Value): {final_metric:.4f} | Total Reward: {total_reward:.1f}")

        # Check if this run is better than the previous best
        if final_metric > best_score:
            best_score = final_metric
            best_params = {'learning_rate': alpha, 'discount_factor': gamma}
            best_q_table = q_table.copy() # Save a copy of the best Q-Table

    print("\n========================================================")
    print("✨ TUNING COMPLETE ✨")
    print(f"BEST PARAMETERS FOUND:")
    print(f"  Learning Rate (Alpha): {best_params.get('learning_rate')}")
    print(f"  Discount Factor (Gamma): {best_params.get('discount_factor')}")
    print(f"  Best Avg Max Q-Value: {best_score:.4f}")
    print("========================================================")

    # Save the best Q-Table locally
    np.save(MODEL_FILENAME, best_q_table)
    print(f"Best Q-Table saved locally as: {MODEL_FILENAME}")

    # Attempt to save to Google Drive (Placeholder)
    save_to_google_drive_placeholder(MODEL_FILENAME)


# --- 4. Google Drive Placeholder Function ---

def save_to_google_drive_placeholder(filepath):
    """
    Saves the file to Google Drive. NOTE: This is a placeholder function.
    You MUST install and configure Google Drive API/pydrive on your local machine
    and replace the comments below with actual API code to make this work.
    """
    print("\n--- Google Drive Save Attempt ---")

    # --------------------------------------------------------------------------
    # NOTE FOR USER:
    # This section requires external libraries (e.g., 'pip install pydrive')
    # and Google API credentials (client_secrets.json).
    # Since this environment is restricted, the actual save code is commented out.
    # --------------------------------------------------------------------------

    try:
        # Example of the logic you would add here (DO NOT UNCOMMENT IN THIS ENVIRONMENT):
        # from pydrive.auth import GoogleAuth
        # from pydrive.drive import GoogleDrive
        # gauth = GoogleAuth()
        # gauth.LocalWebserverAuth() # Authenticate
        # drive = GoogleDrive(gauth)
        # file_list = drive.ListFile({'q': "'root' in parents and trashed=false"}).GetList()
        # file = drive.CreateFile({'title': os.path.basename(filepath)})
        # file.SetContentFile(filepath)
        # file.Upload()

        print(f"File '{filepath}' saved locally. \n"
              f"!!! WARNING: Google Drive API is not active in this environment. \n"
              f"!!! Please run this script on your local machine and uncomment/add \n"
              f"!!! the necessary API code inside this function to upload to Drive.")

    except Exception as e:
        print(f"Error during Google Drive save simulation: {e}")
        print("Please ensure Google Drive API setup is correct on your machine.")

# --- 5. Main Execution ---

if __name__ == "__main__":
    grid_search_tuning()

    print("\nFINAL STEP: Copy the 'nao_emotion_qtable_tuned.npy' file to your Webots controller folder for deployment!")
```
