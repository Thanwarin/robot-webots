# Notebook Summary:

- Converts FER2013 CSV dataset into separate image files organized in folders by emotion class (0–6) and dataset split (train, val, test).
- Reads pixel values, reshapes to 48×48 grayscale images, and saves them as .png.
- Useful for preparing datasets for CNN training or other image-based emotion recognition models.

```python
import pandas as pd
import numpy as np
import os
import cv2
from tqdm import tqdm

# ===========================
# Paths
# ===========================
base_path = "/content/drive/MyDrive/emotion_dataset"
train_path = f"{base_path}/train"
val_path   = f"{base_path}/val"
test_path  = f"{base_path}/test"

os.makedirs(train_path, exist_ok=True)
os.makedirs(val_path, exist_ok=True)
os.makedirs(test_path, exist_ok=True)

# Create class folders (0–6)
for c in range(7):
    os.makedirs(f"{train_path}/{c}", exist_ok=True)
    os.makedirs(f"{val_path}/{c}", exist_ok=True)
    os.makedirs(f"{test_path}/{c}", exist_ok=True)

# ===========================
# Load FER2013 CSV
# ===========================
data = pd.read_csv("/content/fer2013.csv")

def save_image(row, split, idx):
    pixels = np.array(row["pixels"].split(), dtype=np.uint8)
    img = pixels.reshape(48, 48)
    emotion = row["emotion"]

    out_path = f"{base_path}/{split}/{emotion}/{emotion}_{idx}.png"
    cv2.imwrite(out_path, img)

# ===========================
# Export Images
# ===========================
print("Saving images...")

for idx, row in tqdm(data.iterrows(), total=len(data)):
    usage = row["Usage"]

    if usage == "Training":
        save_image(row, "train", idx)
    elif usage == "PublicTest":
        save_image(row, "val", idx)
    elif usage == "PrivateTest":
        save_image(row, "test", idx)

print("Export complete!")
```
